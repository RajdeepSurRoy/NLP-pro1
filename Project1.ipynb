```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import nltk
import string
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

```

```python
from operator import pos
def analyze(text):

    print("\n" + "#" * 30)
    print("NLP LEMMATIZER AND Tagger")
    print("#" * 30)
    print(f"\n Input text")
    print(text)

    tokens = word_tokenize(text)
    print(f"\n Tokenized({len(tokens)}) words")
    print(tokens)
    pos_tags = nltk.pos_tag(tokens)
    print(f"\n POS tags")
    print(f"{'Word':<12}{'Tag':<10}")
    print("-" * 30)
    for word,tag in pos_tags:
        print(f"{word:<12} {tag:<10}")
        cleaned = []
        removed = []

    for word in tokens:
        if word.lower() in stop_words:
            removed.append(word)
            continue
        if not word.isalpha():
            removed.append(word)
            continue
        cleaned.append(word)
        print(f"\n Removed({len(removed)}) words")
        print(removed)
        print(f"\n Cleaned({len(cleaned)}) words")
        print(cleaned)
        lemmatized = [lemmatizer.lemmatize(word.lower(), pos ='v')
                       for word in cleaned]
    print(f"\n Word Transformations:")
    print(f"{'Original':<15} {'Lemmatized'}")
    print("-" * 30)
    for original, lemma in zip(cleaned, lemmatized):
        if original.lower() != lemma:
            print(f"{original:<15} →  {lemma}")
        else:
            print(f"{original:<15} →  (no change)")


    print(f"\n Final Keywords:")
    print("→ " + ", ".join(lemmatized))


    print(f"\n Stats:")
    print(f"Words you typed    : {len(tokens)}")
    print(f"Words removed      : {len(removed)}")
    print(f"Keywords remaining : {len(cleaned)}")
    print(f"Unique lemmas      : {len(set(lemmatized))}")
    print("=" * 45)



```

```python
print("=" * 45)
print(" WELCOME TO NLP LEMMATIZER tagger")
print("=" * 45)
print("Type any sentence to analyze!")
print("Type 'quit' to exit\n")

while True:

    user_text = input(" Enter your text: ")

    if user_text.lower() == 'quit':
        print("\n Goodbye!")
        break

    if len(user_text.strip()) == 0:
        print("Please type something!\n")
        continue

    if len(user_text.split()) < 2:
        print("Please type at least 2 words!\n")
        continue

    analyze(user_text)

    print("Try another or type quit!\n")
```
